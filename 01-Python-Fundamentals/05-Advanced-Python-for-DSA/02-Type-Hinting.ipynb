## 📖 Theory & Concepts

Python is **dynamically typed**, meaning you don't *have* to declare variable types (like `int x = 5` in Java or C++). While this makes Python fast to write, it makes large codebases incredibly hard to read and debug. 

**Type Hinting** (introduced in PEP 484) allows you to explicitly state what type of data a variable, argument, or return value *should* be. 

> [!IMPORTANT]  
> **Type Hints do NOT enforce types at runtime!** Python will not crash if you pass a string to a function expecting an int. However, your IDE (like VS Code or PyCharm) and linters (like `mypy`) will flag it as an error, saving you hours of debugging.

## 💻 1. Basic Type Hinting
You can hint basic variables, function parameters, and return types using a colon `:` and an arrow `->`.


```python
# Hinting Variables (Optional but good practice)
age: int = 25
name: str = "Siva"
is_active: bool = True
price: float = 19.99

# Hinting Functions (Highly Recommended)
def greet(user_name: str, greeting_count: int) -> str:
    return f"Hello {user_name}! " * greeting_count

print(greet("Siva", 2))
```

## 💻 2. Hinting Collections (Lists, Dicts, Sets)
In older Python versions, you had to import `List` and `Dict` from the `typing` module. In Python 3.9+, you can use the built-in lowercase types directly.


```python
# Modern Python 3.9+ Syntax
scores: list[int] = [95, 80, 100]
student_grades: dict[str, str] = {"Siva": "A", "Rahul": "B"}
unique_ids: set[int] = {101, 102, 103}
coordinates: tuple[float, float] = (37.7749, -122.4194)

def process_scores(scores_list: list[int]) -> float:
    return sum(scores_list) / len(scores_list)

print("Average Score:", process_scores(scores))
```

## 💻 3. Union and Optional Types
Sometimes a variable can be more than one type. For example, a function might return an `int` or `None` if something wasn't found.

In Python 3.10+, you can use the pipe `|` operator for this.


```python
from typing import Optional, Union

# Union: Can be int OR float
def double_number(num: Union[int, float]) -> Union[int, float]:
    return num * 2

# Optional: Can be int OR None (Equivalent to Union[int, None])
def find_user_id(username: str) -> Optional[int]:
    database = {"siva": 1, "rahul": 2}
    return database.get(username) # Returns None if not found

# --- MODERN PYTHON 3.10+ SYNTAX (Preferred) ---
def modern_double(num: int | float) -> int | float:
    return num * 2

def modern_find(username: str) -> int | None:
    database = {"siva": 1, "rahul": 2}
    return database.get(username)

print("Found User ID:", modern_find("siva"))
print("Missing User ID:", modern_find("amit"))
```

## 🌍 Real-world Example (Type Hinting in DSA)
In Data Structures and Algorithms, you will constantly create custom Node classes for Linked Lists and Trees. Type hinting makes these structures instantly understandable.

```python
class ListNode:
    def __init__(self, val: int = 0, next_node: 'ListNode | None' = None):
        self.val: int = val
        # Notice we use a string 'ListNode | None' because the class 
        # is not fully defined yet. This is called a "forward reference".
        self.next: ListNode | None = next_node 

def print_linked_list(head: ListNode | None) -> None:
    curr = head
    while curr:
        print(curr.val, end=" -> ")
        curr = curr.next
    print("None")

# Create and print a Linked List: 1 -> 2 -> 3 -> None
node3 = ListNode(3)
node2 = ListNode(2, node3)
head = ListNode(1, node2)

print_linked_list(head)
```

## 🧠 Interview Questions
1. **Does Python enforce type hints at runtime?**
   *Answer: No, Python completely ignores them at runtime. They are strictly for developers, IDEs, and static analysis tools like `mypy`.*
2. **Why do we sometimes need to put type hints inside quotes, like `'ListNode'`?**
   *Answer: Because Python evaluates code top-to-bottom. If you reference a class inside its own `__init__` method, the class isn't fully created yet. The quotes tell Python to evaluate the type later (Forward Reference). Note: In modern Python, you can use `from __future__ import annotations` to avoid needing quotes.*

## ⚠️ Common Mistakes
| Bad Code | Good Code (Pythonic) |
|----------|----------------------|
| `def sum(a, b): return a + b` | `def sum(a: int, b: int) -> int: return a + b` |
| `def find(lst: list):` | `def find(lst: list[int]):` (Always specify what is INSIDE the list) |

## 🔑 Key Takeaways
- Every function you write from now on should have type hints for its arguments and return value.
- Use `list[int]`, `dict[str, int]` to specify what is inside collections.
- Use `| None` or `Optional[...]` if a variable can legitimately be `None`.
- Type hints make your code instantly readable, which FAANG interviewers love to see.
