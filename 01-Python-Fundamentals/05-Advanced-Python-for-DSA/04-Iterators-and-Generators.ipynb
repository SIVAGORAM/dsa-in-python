## 📖 Theory & Concepts

Have you ever wondered how `for i in range(1000000000)` runs instantly without crashing your computer? If it created a list of a billion numbers, your RAM would instantly fill up and crash.

It doesn't crash because `range()` is **lazy**. It generates one number at a time, hands it to the loop, and then forgets it. This concept of "lazy evaluation" is the core of **Iterators** and **Generators**.

---
## 💻 1. Iterable vs Iterator

- **Iterable:** Something you can loop over (like a List, String, or Tuple).
- **Iterator:** The machine doing the looping. It remembers its state and knows how to get the `next()` value.

You convert an Iterable into an Iterator using the `iter()` function.


```python
my_list = [10, 20, 30] # This is an Iterable

# my_list.next() # Error! Lists don't have a next() method.

my_iterator = iter(my_list) # Now it's an Iterator

print(next(my_iterator)) # 10
print(next(my_iterator)) # 20
print(next(my_iterator)) # 30
# print(next(my_iterator)) # This would throw a StopIteration Exception!
```

---
## 💻 2. Generators & The `yield` Keyword

An Iterator requires you to write a whole Class with `__iter__` and `__next__` methods (which is tedious). 
A **Generator** is a shortcut. It is a function that uses the `yield` keyword instead of `return`.

> [!TIP]  
> `return` destroys the function's local variables and exits completely. 
> `yield` just *pauses* the function, hands a value back, and waits to be resumed.


```python
def count_up_to(max_val):
    count = 1
    while count <= max_val:
        yield count # PAUSE here, and hand 'count' to the caller
        count += 1

# Create the generator
counter = count_up_to(3)

print("First call:", next(counter))
print("Second call:", next(counter))
print("Third call:", next(counter))

# Or just use it in a for loop (the loop handles StopIteration automatically)
print("Using a for loop:")
for num in count_up_to(3):
    print(num)
```

## 💻 3. Generator Expressions (Space Complexity Savior)

In the Comprehensions notebook, you learned List Comprehensions. If you use parenthesis `()` instead of brackets `[]`, you create a **Generator Expression**.

This takes **O(1) Space Complexity** instead of **O(N) Space Complexity**.


```python
import sys

# List Comprehension: Creates ALL 1,000,000 numbers in RAM immediately
list_comp = [x ** 2 for x in range(1_000_000)]
print(f"List memory size: {sys.getsizeof(list_comp)} bytes")

# Generator Expression: Creates NOTHING in RAM yet. Just knows HOW to generate them.
gen_expr = (x ** 2 for x in range(1_000_000))
print(f"Generator memory size: {sys.getsizeof(gen_expr)} bytes")

# You can still iterate over the generator just like the list!
print("First value of gen:", next(gen_expr))
```

## 🌍 Real-world Example (Infinite Streams)
Generators are perfect for representing data streams that literally never end (like a live feed of stock prices or reading a massive 500GB log file line-by-line).

```python
def even_numbers():
    """An INFINITE generator of even numbers"""
    n = 0
    while True:
        yield n
        n += 2

evens = even_numbers()

# We can just grab the first 5 without crashing the computer
for _ in range(5):
    print(next(evens))
```

## 🧠 Interview Questions
1. **What is the difference between `return` and `yield`?**
   *Answer: `return` terminates the function entirely and destroys its local state. `yield` pauses the function, returns a value, and saves the local state so the function can be resumed later.*
2. **When should you use a Generator instead of a List Comprehension?**
   *Answer: When you are dealing with a massive dataset (e.g., millions of rows) and you only need to process one item at a time. A generator keeps Space Complexity at O(1), whereas a List Comprehension takes O(N) memory.*

## 🔑 Key Takeaways
- Iterables (Lists, Strings) store data. Iterators (Generators) *produce* data.
- `yield` creates a Generator.
- Generators are highly memory efficient (O(1) space).
- Generators can only be looped over **once**.
