## 📖 Theory & Concepts

When solving Data Structure and Algorithm problems, you are judged on two things: **Time Complexity (Big O)** and **Space Complexity**.

In Python, built-in functions like `in`, `append()`, and `pop()` feel like magic because they happen instantly on small data. However, under the hood, some of these operations take **O(1)** (instant) time, while others take **O(N)** (linear) time.

> [!WARNING]  
> If you write a `for` loop that runs N times, and inside that loop you call a built-in Python function that takes O(N) time, **your algorithm is secretly O(N²)**. This is the #1 reason candidates fail FAANG interviews.

---
## 💻 1. List Complexity (The Danger Zone)

Lists are implemented as contiguous arrays in memory.

| Operation | Syntax | Time Complexity | Why? |
|-----------|--------|-----------------|------|
| **Append** | `l.append(x)` | **O(1)** | Just adds to the end of the pre-allocated memory. |
| **Pop End** | `l.pop()` | **O(1)** | Just removes the last item. |
| **Index Access**| `l[i]` | **O(1)** | Direct memory access via offset. |
| **Pop First** | `l.pop(0)` | **O(N)** ❌ | Forces Python to shift all N-1 remaining elements left. |
| **Insert First**| `l.insert(0, x)`| **O(N)** ❌ | Forces Python to shift all N elements right. |
| **Search** | `x in l` | **O(N)** ❌ | Has to check every single element one by one. |


````python
import time

n = 100_000
test_list = list(range(n))

# BENCHMARK: Pop from the END O(1)
start = time.time()
while test_list:
    test_list.pop()
end = time.time()
print(f"O(1) pop() took: {end - start:.5f} seconds")

# BENCHMARK: Pop from the FRONT O(N) -> Total O(N^2)
test_list = list(range(n))
start = time.time()
while test_list:
    test_list.pop(0)
end = time.time()
print(f"O(N) pop(0) took: {end - start:.5f} seconds")
````

## 💻 2. Set and Dictionary Complexity (The Savior)

Sets and Dicts are implemented using **Hash Tables**.

| Operation | Syntax | Time Complexity | Why? |
|-----------|--------|-----------------|------|
| **Add/Set** | `s.add(x)` / `d[k] = v` | **O(1)** | Hashes the key and jumps directly to the memory bucket. |
| **Search** | `x in s` / `k in d` | **O(1)** ✅ | Instant lookup via hash! (Compare to List's O(N)). |
| **Delete** | `s.remove(x)` / `del d[k]`| **O(1)** | Instant deletion. |

> [!TIP]  
> If you need to repeatedly check `if x in collection`, **always** convert the collection to a `set` first.


````python
import time

n = 10_000_000
test_list = list(range(n))
test_set = set(test_list)
target = n - 1 # Searching for the very last element

# BENCHMARK: List Search O(N)
start = time.time()
found = target in test_list
end = time.time()
print(f"O(N) List 'in' search took: {end - start:.5f} seconds")

# BENCHMARK: Set Search O(1)
start = time.time()
found = target in test_set
end = time.time()
print(f"O(1) Set 'in' search took: {end - start:.5f} seconds")
````

## 💻 3. String Concatenation Complexity
Strings in Python are **immutable**. You cannot change a string; you can only create a new one.

If you build a string using `+` inside a loop, Python has to create a brand new string and copy all the characters over every single iteration. This makes `s += "x"` an **O(N)** operation inside a loop, leading to **O(N²)** total time.


````python
import time

n = 100_000

# BAD: String Concatenation O(N^2)
start = time.time()
s = ""
for _ in range(n):
    s += "a"
end = time.time()
print(f"O(N^2) String concatenation (+): {end - start:.5f} seconds")

# GOOD: List Append + Join O(N)
start = time.time()
chars = []
for _ in range(n):
    chars.append("a")
s = "".join(chars)
end = time.time()
print(f"O(N) List append + .join(): {end - start:.5f} seconds")
````

## 🌍 Real-world Example (Silent Performance Killer)
Imagine you are given a list of 100,000 bad words, and a document of 1,000,000 words. You need to count how many bad words are in the document.

````python
def count_bad_words_bad(document: list[str], bad_words: list[str]) -> int:
    # Time Complexity: O( len(document) * len(bad_words) )
    count = 0
    for word in document:
        if word in bad_words: # <--- O(N) SEARCH HIDDEN HERE!
            count += 1
    return count

def count_bad_words_good(document: list[str], bad_words: list[str]) -> int:
    # Time Complexity: O( len(document) + len(bad_words) )
    bad_set = set(bad_words) # O(N) to build the set once
    count = 0
    for word in document:
        if word in bad_set: # <--- O(1) SEARCH!
            count += 1
    return count
````

## 🧠 Interview Questions
1. **I have a list of items and I need to constantly remove items from the front of it. What should I do?**
   *Answer: Do not use `list.pop(0)`. Import `collections.deque` and use `deque.popleft()` instead.*
2. **Why does inserting into a set take O(1) time?**
   *Answer: Sets use Hash Tables. When you insert a value, Python computes its hash, which maps directly to a memory address (a bucket). It doesn't need to iterate over existing elements.*

|----------|----------------------|
| `if item in my_huge_list:` | `if item in my_huge_set:` |
| `my_list.pop(0)` | `my_deque.popleft()` |
| `s = ""`<br>`for char in arr: s += char` | `s = "".join(arr)` |

---
## 🔑 Key Takeaways
- `list.append()` and `list.pop()` are fast (O(1)).
- `list.insert(0)` and `list.pop(0)` are extremely slow (O(N)).
- Searching in a List is slow (O(N)).
- Searching in a Set or Dict is instant (O(1)).
- Never build large strings using `+=` in a loop. Always append to a list and use `"".join()`.
