## 📖 Theory & Concepts

Python's built-in `list` and `dict` are fantastic, but they have limitations in Data Structures and Algorithms:
1. **`list.pop(0)` is O(N)**: Removing the first element of a list forces Python to shift every other element in memory. If you do this in a loop, your O(N) algorithm becomes O(N²).
2. **Frequency Counting is tedious**: Writing loops to count character occurrences takes too many lines of code.
3. **`KeyError` in Graphs**: When building adjacency lists (e.g., `graph[u].append(v)`), if `u` doesn't exist yet, it throws a `KeyError`.

The `collections` module solves all of these problems.

---
## 💻 1. `deque` (Double-Ended Queue)
A `deque` (pronounced "deck") allows O(1) appends and pops from **both** ends. 

> [!IMPORTANT]  
> **Never use a list as a Queue in a FAANG interview.** Always use `from collections import deque`.


```python
from collections import deque

# Initialize a deque
queue = deque([1, 2, 3])

# O(1) Append to right (like a normal list)
queue.append(4)

# O(1) Append to left (IMPOSSIBLE with normal list)
queue.appendleft(0)

print("Deque after appends:", queue)

# O(1) Pop from right
right_val = queue.pop()

# O(1) Pop from left (Used in Breadth-First Search!)
left_val = queue.popleft()

print(f"Popped Right: {right_val}, Popped Left: {left_val}")
print("Deque after pops:", queue)
```

## 💻 2. `Counter`
A `Counter` is a subclass of dictionary designed specifically for counting hashable objects. It turns an O(N) manual loop into a single, highly optimized line of C-code.


```python
from collections import Counter

text = "abracadabra"
freq = Counter(text)

print("Frequency Map:", freq)
print("Count of 'a':", freq['a'])
print("Count of 'z' (doesn't exist):", freq['z']) # Returns 0 instead of KeyError!

# Find the 2 most common characters
print("Most common:", freq.most_common(2))
```

## 💻 3. `defaultdict`
A `defaultdict` never raises a `KeyError`. If you try to access a key that doesn't exist, it automatically creates it using a default factory function (like `int`, `list`, or `set`).

> [!TIP]  
> `defaultdict(list)` is the industry standard for building Graph Adjacency Lists.


```python
from collections import defaultdict

# Example 1: Default to Integer (0)
scores = defaultdict(int)
scores["Siva"] += 10 # Normally this would throw KeyError because "Siva" doesn't exist
print("Scores:", scores)

# Example 2: Default to List [] (Graph Adjacency List)
edges = [("A", "B"), ("A", "C"), ("B", "C")]
graph = defaultdict(list)

for u, v in edges:
    graph[u].append(v)
    # Without defaultdict, you'd have to write: 
    # if u not in graph: graph[u] = []
    # graph[u].append(v)

print("Graph Adjacency List:", dict(graph))
```

## 🧠 Interview Questions
1. **Why is `list.pop(0)` O(N) but `deque.popleft()` is O(1)?**
   *Answer: A list in Python is implemented as a contiguous array in memory. Removing the first item requires shifting all N-1 remaining items to the left. A `deque` is implemented as a doubly-linked list (or a block-linked list in CPython), so removing the head only requires updating a pointer.*
2. **When should you NOT use a `deque`?**
   *Answer: When you need random access. Accessing the middle of a `deque` (e.g., `queue[500]`) is O(N) because it has to traverse the linked list, whereas `list[500]` is O(1).*
