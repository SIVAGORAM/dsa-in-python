## 1. What is a Data Type?
Imagine you are packing for a trip. You put water in a bottle, clothes in a suitcase, and your laptop in a padded bag. You use different containers for different things because they have different properties.

In Python, a **Data Type** is the "container type" for your data. It tells the Python interpreter:
1. How much **RAM (memory)** to allocate.
2. What **operations** are allowed (e.g., you can multiply numbers, but you cannot multiply a string by another string).


```python
age = 22             # Python knows this is a Number
name = "Siva"        # Python knows this is Text
is_hired = True      # Python knows this is a Yes/No boolean
```

## 2. Numeric Types (`int`, `float`) & Memory Caching

### Integer (`int`)
Whole numbers, positive or negative, without decimals.


```python
apples = 10
temperature = -5
print(type(apples))
```

### 🧠 FAANG Deep Dive: Integer Caching (Interning)
To optimize speed and memory, Python pre-loads all integers from **-5 to 256** when it starts up. If you create two variables with the same value in this range, Python points them to the **exact same memory address**!


```python
x = 100
y = 100
print(x is y)  # True! Both point to the same cached 100 in RAM.

a = 300
b = 300
print(a is b)  # False! 300 is outside the cache, so Python creates two separate objects.
```

### Float (`float`)
Numbers with a decimal point. Python floats are implemented using C's `double` in the backend, meaning they have high precision but can suffer from floating-point arithmetic issues.


```python
price = 19.99
print(type(price))
```

## 3. Text Sequence (`str`)
Strings are sequences of characters. They can be enclosed in single `' '`, double `" "`, or triple `''' '''` quotes.


```python
city = "Kurnool"
print(type(city))
```

### 🧠 FAANG Deep Dive: String Interning
Similar to numbers, Python caches short strings (especially those that look like variable names, containing only letters/numbers/underscores). This makes comparing dictionary keys incredibly fast.

## 7. 🔥 Mutable vs Immutable (CRITICAL)
This is the **most important concept** in Python memory management.

| Immutable (Cannot be changed in RAM) | Mutable (Can be changed in RAM) |
|--------------------------------------|---------------------------------|
| `int`, `float`, `bool`               | `list`                          |
| `str` (String)                       | `dict` (Dictionary)             |
| `tuple`                              | `set`                           |
| `NoneType`                           |                                 |

**What does this mean?**
If you create a string `name = "Siva"` and then do `name = "Siva Goram"`, Python **did not modify** the original string. It destroyed the old string in RAM and created a brand new one!
However, if you have a list `x = [1, 2]`, and you do `x.append(3)`, Python modifies the exact same list in RAM without moving it.


```python
# PROOF:
name = "Siva"
print(id(name))  # Look at the memory address
name += " Goram"
print(id(name))  # The address CHANGED! Strings are immutable.

my_list = [1, 2]
print(id(my_list))
my_list.append(3)
print(id(my_list))  # The address is EXACTLY THE SAME! Lists are mutable.
```

## 8. Type Checking
You can check a variable's type using `type()`. However, in professional code, we use `isinstance()` because it handles inheritance properly.


```python
age = 22
print(type(age) == int)  # True
print(isinstance(age, int))  # True (Best Practice!)
```

## 9. Type Casting (Conversion)
Sometimes you need to convert data from one type to another.

### Implicit Casting (Automatic)
Python automatically converts lower data types to higher data types to prevent data loss.


```python
x = 10     # int
y = 2.5    # float
z = x + y  # Python automatically promotes 'x' to a float before adding!
print(z, type(z))
```

### Explicit Casting (Manual)
You force Python to convert a type using functions like `int()`, `float()`, `str()`, `bool()`, `list()`, etc.


```python
# String to Int
age_str = "25"
age_int = int(age_str)

# Int to String
salary = 50000
salary_str = str(salary)

# Float to Int (WARNING: You lose the decimals!)
pi = 3.1415
pi_int = int(pi)
print(pi_int)  # Outputs 3
```

## 10. 🏆 FAANG Interview Questions

1. **What is the difference between mutable and immutable data types? Give examples.**
   *Answer:* Mutable objects (Lists, Dicts, Sets) can be modified in-place in memory. Immutable objects (Ints, Strings, Tuples) cannot be changed once created; any "modification" actually creates a brand new object in memory.

2. **Why does `a = 256; b = 256; print(a is b)` output True, but `x = 257; y = 257; print(x is y)` output False?**
   *Answer:* Integer caching. Python pre-allocates an array of integers from -5 to 256. Variables assigned to values in this range point to the same cached memory address. Outside this range, Python creates new distinct objects.

3. **Can a Tuple be modified? What if a Tuple contains a List?**
   *Answer:* A Tuple itself is immutable, meaning its memory pointers cannot change. You cannot add, remove, or reassign elements. However, if an element inside the tuple is a mutable object (like a List), you *can* modify the contents of that List!

4. **Is `True + True` valid Python? What does it output?**
   *Answer:* Yes. `bool` is a subclass of `int` in Python. `True` evaluates to `1`, so the output is `2`.
