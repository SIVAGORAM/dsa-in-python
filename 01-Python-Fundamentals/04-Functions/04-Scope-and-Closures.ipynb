## 📖 Theory & Concepts

**Scope** defines the visibility and lifetime of a variable. If you declare a variable inside a function, can you use it outside? No. This is because of Scope.

Python resolves variables using the **LEGB Rule** (searches in this exact order):
1. **Local (L):** Variables defined inside the current function.
2. **Enclosing (E):** Variables in the local scope of enclosing (nested) functions.
3. **Global (G):** Variables defined at the top-level of the script/module.
4. **Built-in (B):** Names preassigned in Python (`len`, `print`, etc.).

---
## 💻 1. Local and Global Scope
A variable declared inside a function is **Local**. A variable declared outside all functions is **Global**.

```python
x = "Global X" # Global Scope

def test_scope():
    y = "Local Y" # Local Scope
    print("Inside function, can read global:", x)
    print("Inside function, can read local:", y)

test_scope()
print("Outside function, can read global:", x)
# print(y) # THIS WOULD CAUSE A NameError! 'y' is destroyed after the function finishes.
```

---
## 💻 2. The `global` Keyword
By default, if you assign a value to a variable inside a function, Python treats it as a *new* local variable. To modify a global variable inside a function, you must use the `global` keyword.

```python
counter = 0

def increment():
    # Without the next line, 'counter += 1' would throw an UnboundLocalError
    global counter 
    counter += 1

increment()
increment()
print("Global Counter:", counter)
```

---
## 💻 3. Enclosing Scope & `nonlocal` Keyword (CRUCIAL FOR DSA)
In Python, you can define functions *inside* other functions (nested functions). The outer function's scope is the **Enclosing Scope** for the inner function.

If the inner function wants to modify a variable from the outer function, it must use the `nonlocal` keyword. **You will use this constantly in Depth First Search (DFS) and Tree traversal algorithms.**

```python
def outer_function():
    outer_var = "I am Outer"
    
    def inner_function():
        nonlocal outer_var # Tells Python to modify the variable in the enclosing scope
        outer_var = "Outer was modified by Inner!"
    
    print("Before Inner call:", outer_var)
    inner_function()
    print("After Inner call:", outer_var)

outer_function()
```

> [!IMPORTANT]  
> **Why `nonlocal` matters in DSA:** When writing recursive helper functions (like `dfs()`), you often need to track a global maximum or a counter. Instead of using a messy `global` variable, you initialize the tracker in the main function and use `nonlocal` inside the recursive helper.

---
## 💻 4. Closures
A **Closure** occurs when an inner function remembers the state of its enclosing environment even after the outer function has finished executing.

```python
def multiplier_factory(n):
    # 'n' is in the enclosing scope
    def multiplier(x):
        return x * n
    return multiplier # We return the inner function ITSELF, not its result

# double and triple are Closures.
# They "remember" that n was 2 and 3 respectively.
double = multiplier_factory(2)
triple = multiplier_factory(3)

print("Double of 5:", double(5))
print("Triple of 5:", triple(5))
```

---
## 🌍 Real-world Example
You want to build a simple hit-counter for a web page. Instead of using a class, you can use a closure to keep the count private and persistent.

```python
def create_hit_counter():
    hits = 0
    def increment_hit():
        nonlocal hits
        hits += 1
        return hits
    return increment_hit

button_click_tracker = create_hit_counter()
print("Click 1:", button_click_tracker())
print("Click 2:", button_click_tracker())
print("Click 3:", button_click_tracker())
```

## 🧠 Interview Questions
1. **What is the difference between `global` and `nonlocal`?**
   *Answer: `global` is used to modify variables defined at the top (module) level. `nonlocal` is used to modify variables defined in the nearest enclosing (outer) function.*
2. **Can you read a global variable inside a function without the `global` keyword?**
   *Answer: Yes. You only need the `global` keyword if you plan to ASSIGN/MODIFY the variable (`x = 5`). If you just want to read it (`print(x)`), Python's LEGB rule finds it automatically.*

## ⚠️ Common Mistakes
| Bad Code | Good Code (Pythonic) |
|----------|----------------------|
| Heavy reliance on `global` variables to pass state around algorithms. | Using `nonlocal` inside a nested helper function, or passing the state as a function parameter. |

---
## 🔑 Key Takeaways
- Remember **LEGB**: Local -> Enclosing -> Global -> Built-in.
- `global` allows modification of script-level variables.
- `nonlocal` allows modification of outer-function variables (essential for recursive DSA helpers).
- Closures allow functions to "remember" the state of their enclosing scope.
